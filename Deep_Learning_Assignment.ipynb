Question 1: What is Deep Learning? Briefly describe how it evolved and how it differs from traditional machine learning.

Answer :
Deep Learning is a subfield of Artificial Intelligence (AI) and Machine Learning (ML) that uses artificial neural networks with multiple layers to learn patterns and features automatically from large amounts of data. It is inspired by the structure and working of the human brain. Deep Learning is widely used in image recognition, speech recognition, natural language processing, and self-driving cars.

2. Evolution of Deep Learning
- 1940s–1950s: The basic concept of artificial neurons was introduced.
- 1980s: Neural networks and backpropagation became important research areas.
- 2000s: Improvements in computing power and availability of large datasets supported the development of deep learning.
- 2010s: Deep learning achieved major success in image recognition, speech processing, and other AI applications.
- Present: Deep learning is widely used in generative AI, robotics, healthcare, and autonomous systems.

Deep Learning is an advanced form of Machine Learning that uses multilayer neural networks to automatically learn complex patterns from data. It has evolved due to improvements in algorithms, computing power, and data availability, making it an important technology in modern Artificial Intelligence.

Question 2: Explain the basic architecture and functioning of a Perceptron. What are its limitations?

Answer :
A Perceptron is the simplest type of artificial neural network used for binary classification. It is a single-layer neural network that takes input values, assigns weights to them, adds a bias, and applies an activation function to produce an output.

A perceptron consists of the following components:
1. Input: Receives input values such as \(x_1, x_2, x_3\).
2. Weights: Assigns importance to each input.
3. Bias: Adjusts the output of the perceptron.
4. Summation Function: Calculates the weighted sum of inputs.
5. Activation Function: Determines the final output based on the calculated sum.

The working of a perceptron involves the following steps:
1. The perceptron receives input values.
2. Each input is multiplied by its corresponding weight.
3. All weighted inputs are added together with the bias.
4. The activation function processes the sum.
5. The perceptron produces the final output, usually 0 or 1.
6. During training, weights and bias are updated based on prediction errors.

Limitations of a Perceptron
1. Linear Classification: It can classify only linearly separable data.
2. XOR Problem: It cannot solve the XOR problem using a single perceptron.
3. Limited Complexity: It cannot learn complex patterns and relationships.
4. Single-Layer Structure: It has no hidden layers to learn advanced features.
5. Limited Output: It is mainly suitable for simple binary classification tasks.
6. Data Dependency: Its performance depends on the quality of the training data.

A perceptron is the basic building block of artificial neural networks. It performs classification using inputs, weights, bias, and an activation function. However, due to its inability to solve nonlinear problems, more advanced neural networks with multiple layers are used for complex tasks.

Question 3: Describe the purpose of activation function in neural networks. Compare Sigmoid, ReLU, and Tanh functions.

Answer :-
An activation function is a mathematical function used in an artificial neural network to determine the output of a neuron. It introduces non-linearity into the network, allowing it to learn complex patterns and relationships from data. Without activation functions, a neural network would behave like a linear model, even with multiple layers.

Purpose of Activation Function
1. Introduces Non-linearity: Helps neural networks learn complex relationships.
2. Controls Neuron Output: Determines whether a neuron should be activated.
3. Improves Learning: Helps the network learn patterns from training data.
4. Transforms Output: Converts the weighted sum of inputs into a suitable output range.
5. Supports Deep Learning: Enables neural networks to solve complex problems such as image recognition and speech processing.

3. Types of Activation Functions

A. Sigmoid Function :

The Sigmoid function converts input values into a range between 0 and 1. It is commonly used in the output layer for binary classification.

B. ReLU (Rectified Linear Unit)

ReLU is a widely used activation function in hidden layers of deep neural networks. It returns zero for negative inputs and the input itself for positive inputs.

C. Tanh (Hyperbolic Tangent)

The Tanh function converts input values into a range between −1 and +1. It is a zero-centered activation function.

Activation functions are essential components of neural networks because they introduce non-linearity and enable the learning of complex patterns. Sigmoid is useful for binary classification outputs, ReLU is widely used in hidden layers because of its efficiency, and Tanh is useful when outputs centered around zero are preferred.

Question 4: What is the difference between Loss function and Cost function in neural networks? Provide examples.

Answer :

A Loss Function is a mathematical function that measures the difference between the actual output and the predicted output for a single training example. It helps the neural network understand how far its prediction is from the correct answer.

A Cost Function measures the overall error of a neural network across the entire training dataset or a batch of training examples. It is generally calculated by averaging the individual loss values.

The loss function measures the error for an individual prediction, whereas the cost function measures the overall error across a dataset or batch. Both are important in neural networks because they help evaluate model performance and guide the optimization process to reduce prediction errors.

Question 5: What is the role of optimizers in neural networks? Compare Gradient
Descent, Adam, and RMSprop.

Answer :
An optimizer is an algorithm used in neural networks to update the weights and biases of the network to minimize the loss function. It improves the model's performance by reducing prediction errors during the training process.

Role of Optimizers in Neural Networks
1. Minimize Loss: Reduces the error between actual and predicted outputs.
2. Update Weights: Adjusts weights and biases during training.
3. Improve Accuracy: Helps the model learn better patterns from data.
4. Speed Up Training: Some optimizers help the model converge faster.
5. Improve Convergence: Helps the model reach a suitable minimum of the loss function.

Types of Optimizers

A. Gradient Descent :
Gradient Descent is an optimization algorithm that updates model parameters in the opposite direction of the gradient of the loss function.

B. Adam (Adaptive Moment Estimation) :
Adam is an advanced optimizer that combines momentum and adaptive learning rates to update model parameters efficiently.

C. RMSprop (Root Mean Square Propagation)
RMSprop is an adaptive optimization algorithm that adjusts the learning rate for each parameter using a moving average of squared gradients.

Optimizers play an important role in training neural networks by minimizing loss and updating weights and biases. Gradient Descent is simple and fundamental, Adam provides adaptive learning rates with momentum, and RMSprop adjusts learning rates based on recent squared gradients. Adam is a common default choice for many deep learning applications.

Question 6: Write a Python program to implement a single-layer perceptron from
scratch using NumPy to solve the logical AND gate.

Answer:

    import numpy as np

    # Input data for AND gate
    X = np.array([[0, 0],
                  [0, 1],
                  [1, 0],
                  [1, 1]])

    # Expected output
    y = np.array([0, 0, 0, 1])

    # Initialize weights and bias
    weights = np.zeros(2)
    bias = 0
    learning_rate = 0.1
    epochs = 10

    # Step activation function
    def activation(x):
        return 1 if x >= 0 else 0

    # Training the perceptron
    for epoch in range(epochs):
        for i in range(len(X)):
            z = np.dot(X[i], weights) + bias
            prediction = activation(z)

            error = y[i] - prediction

            weights += learning_rate * error * X[i]
            bias += learning_rate * error

    # Testing the perceptron
    print("AND Gate Output:")
    for i in range(len(X)):
        z = np.dot(X[i], weights) + bias
        prediction = activation(z)
        print(X[i], "=", prediction)

Question 7: Implement and visualize Sigmoid, ReLU, and Tanh activation functions
using Matplotlib.

Answer :

    import numpy as np
    import matplotlib.pyplot as plt

    # Generate input values
    x = np.linspace(-10, 10, 100)

    # Activation functions
    sigmoid = 1 / (1 + np.exp(-x))
    relu = np.maximum(0, x)
    tanh = np.tanh(x)

    # Create subplots
    plt.figure(figsize=(12, 4))

    # Sigmoid
    plt.subplot(1, 3, 1)
    plt.plot(x, sigmoid)

Question 8: Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy.

Answer :

    import tensorflow as tf
    from tensorflow import keras
    from tensorflow.keras import layers

    # Load MNIST dataset
    (X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

    # Normalize pixel values
    X_train = X_train / 255.0
    X_test = X_test / 255.0

    # Build the neural network
    model = keras.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    # Compile the model
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    # Train the model
    model.fit(X_train, y_train, epochs=5, batch_size=32)

    # Print training accuracy
    loss, accuracy = model.evaluate(X_train, y_train, verbose=0)
    print("Training Accuracy:", accuracy * 100, "%")

    import tensorflow as tf

    from tensorflow.keras import layers

    model = keras.Sequential([
        layers.Flatten(input_shape=(28, 28)),
        layers.Dense(128, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    # Compile model
    model.compile(

        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    # Train model
    history = model.fit(
        X_train, y_train,
        epochs=10,
        batch_size=32,
        validation_split=0.2

    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)

    # Plot accuracy curves

Question 10: You are working on a project for a bank that wants to automatically detect fraudulent transactions. The dataset is large, imbalanced, and contains structured features like transaction amount, merchant ID, and customer location. The goal is to classify each transaction as fraudulent or legitimate.
Explain your real-time data science workflow:

    ● How would you design a deep learning model (perceptron or multilayer NN)?
    ● Which activation function and loss function would you use, and why?
    ● How would you train and evaluate the model, considering class imbalance?
    ● Which optimizer would be suitable, and how would you prevent overfitting?

Answer

1. Introduction
Fraud detection is a binary classification problem in which a deep learning model classifies bank transactions as fraudulent (1) or legitimate (0). Since the dataset is large, structured, and imbalanced, a multilayer neural network can be used to learn complex patterns and identify suspicious transactions.
2. Design of Deep Learning Model
I would use a Multilayer Neural Network (MLP) instead of a single-layer perceptron because it can learn complex relationships between transaction amount, merchant ID, customer location, and other features.
The model consists of:
1. Input layer: Receives transaction features.
2. Hidden layers: Use 64 and 32 neurons with ReLU activation.
3. Output layer: Uses one neuron with Sigmoid activation to predict the probability of fraud.
3. Selection of Activation and Loss Functions
- ReLU: Used in hidden layers because it is computationally efficient and helps learn nonlinear patterns.
- Sigmoid: Used in the output layer because it produces a probability between 0 and 1.
- Binary Cross-Entropy: Used as the loss function because the task has two classes: fraudulent and legitimate.
4. Training and Evaluation with Imbalanced Data
Fraudulent transactions are usually much fewer than legitimate transactions. Therefore, accuracy alone is not sufficient.
- Split the data into training, validation, and test sets using stratified sampling.
- Normalize numerical features and encode categorical features.
- Use class weights to give more importance to fraudulent transactions.
- Evaluate using precision, recall, F1-score, PR-AUC, and a confusion matrix.
- Choose a decision threshold based on the bank's fraud detection requirements.
5. Optimizer and Overfitting Prevention
- Optimizer: Adam is suitable because it adapts learning rates and often converges efficiently.
- Dropout: Randomly disables some neurons during training to reduce overfitting.
- Early stopping: Stops training when validation performance stops improving.
- L2 regularization: Penalizes excessively large weights.
- Class weighting: Helps the model learn from the minority fraud class.


      import numpy as np
      import tensorflow as tf
      from tensorflow import keras
      from tensorflow.keras import layers, regularizers
      from sklearn.datasets import make_classification
      from sklearn.model_selection import train_test_split
      from sklearn.preprocessing import StandardScaler
      from sklearn.metrics import (
          classification_report, confusion_matrix,
          average_precision_score
      )

      # 1. Create an imbalanced sample dataset
      X, y = make_classification(
          n_samples=10000,
          n_features=10,
          n_informative=6,
          n_redundant=2,
          weights=[0.98, 0.02],
          random_state=42
      )

      # 2. Split data using stratified sampling
      X_train, X_test, y_train, y_test = train_test_split(
          X, y, test_size=0.2,
          stratify=y, random_state=42
      )

      # 3. Scale features
      scaler = StandardScaler()
      X_train = scaler.fit_transform(X_train)
      X_test = scaler.transform(X_test)

      # 4. Calculate class weights
      counts = np.bincount(y_train)
      class_weights = {
          0: len(y_train) / (2 * counts[0]),
          1: len(y_train) / (2 * counts[1])
      }

      # 5. Build multilayer neural network
      model = keras.Sequential([
          keras.Input(shape=(X_train.shape[1],)),
          layers.Dense(
              64, activation="relu",
              kernel_regularizer=regularizers.l2(0.001)
          ),
          layers.Dropout(0.3),
          layers.Dense(32, activation="relu"),
          layers.Dropout(0.2),
          layers.Dense(1, activation="sigmoid")
      ])

      # 6. Compile model
      model.compile(
          optimizer="adam",
          loss="binary_crossentropy",
          metrics=[
              keras.metrics.Precision(name="precision"),
              keras.metrics.Recall(name="recall"),
              keras.metrics.AUC(curve="PR", name="pr_auc")
          ]
      )

      # 7. Train model
      early_stop = keras.callbacks.EarlyStopping(
          monitor="val_pr_auc",
          mode="max",
          patience=3,
          restore_best_weights=True
      )

      model.fit(
          X_train, y_train,
          epochs=20,
          batch_size=64,
          validation_split=0.2,
          class_weight=class_weights,
          callbacks=[early_stop],
          verbose=0
      )

      # 8. Predict fraud probabilities
      probabilities = model.predict(X_test, verbose=0).ravel()
      threshold = 0.5
      predictions = (probabilities >= threshold).astype(int)

      # 9. Evaluate model
      print("Confusion Matrix:")
      print(confusion_matrix(y_test, predictions))

      print("\nClassification Report:")
      print(classification_report(
          y_test, predictions,
          target_names=["Legitimate", "Fraud"],
          zero_division=0
      ))

      print("PR-AUC:",
            round(average_precision_score(y_test, probabilities), 4))
